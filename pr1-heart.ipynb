```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

# Step 1: Load Dataset
data = pd.read_csv("Heart.csv")
print("✅ Data loaded successfully!")
print(data)
```

```python
# Step 2: Data Cleaning
# Remove duplicate rows if any
data = data.drop_duplicates()

# Check for missing values
print("\nMissing values before cleaning:\n", data.isnull().sum())
```

```python
# Fill missing values
data['Ca'] = data['Ca'].fillna(data['Ca'].mean())
data['Thal'] = data['Thal'].fillna(data['Thal'].mode()[0])
data.isnull().sum()
```

```python
# Step 4: Data Transformation
# Encode categorical variables if any
for col in data.columns:
    if data[col].dtype == 'object':
        le = LabelEncoder()
        data[col] = le.fit_transform(data[col])
print(data)
```

```python
# Feature Scaling- Standardization (for numeric values)
numeric_cols = ['Age', 'RestBP', 'Chol', 'MaxHR', 'Oldpeak']
scaler = StandardScaler()
data[numeric_cols] = scaler.fit_transform(data[numeric_cols])
print(data)
```

```python
# Step 5: Separate features (X) and target (y)
X = data.drop('AHD', axis=1)
y = data['AHD']
# print(X)
# print(y)
```
